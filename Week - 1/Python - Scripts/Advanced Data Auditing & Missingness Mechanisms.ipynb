#### **Step 6: Data Standardization**

```python
df.head()
```

```python
df['Gender'].value_counts()
```

##### **6.1 Standardize Gender**

```python
df['Gender'] = df['Gender'].replace({
                  'M': 'Male',
                  'male': 'Male',
                  'F': 'Female',
                  'FEMALE': 'Female'
                  })
```

```python
df['Gender'].value_counts()
```

##### **6.2 Standardize Region**

```python
df['Region'].value_counts()
```

```python
df['Region'] = df['Region'].replace({
    'NORTH': 'North',
    'SOUTH': 'South',
    'EAST': 'East',
    'WEST': 'West',
    'CENTRAL': 'Central',
})
```

```python
df['Region'].value_counts()
```

##### **6.3 Standardize Product Category Column**

```python
df['ProductCategory'].value_counts()
```

```python
df['ProductCategory'] = (
    df['ProductCategory']
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    )
```

```python
df['ProductCategory'].value_counts()
```

```python
df['PaymentMethod'].value_counts()
```

```python
df['CustomerSegment'].value_counts()
```

##### **6.4 Standardize the City Column**

```python
df['City'].value_counts()
```

```python
# Capitalize the first letter of each City 
df['City'] = df['City'].str.strip().str.title()
```

```python
df['City'].value_counts()
```

```python
df['SalesRep'].value_counts()
```

##### **6.5 Put Together First and Last Name to Create a New Column Called CustomerName**

```python
df['CustomerName'] = df['FirstName'] + ' ' + df['LastName']

# Place CustomerName in the position of the FirstName column
first_name_index = df.columns.get_loc('FirstName')
df.insert(first_name_index, 'CustomerName', df.pop('CustomerName'))

# Delete the FirstName and LastName columns
df.drop(['FirstName', 'LastName'], axis=1, inplace=True)
df.head(10)
```

#### **Step 7: Summary Statistics for Continuous Data**

```python
summary = df[['Quantity', 'UnitPrice', 'Discount', 'Rating', 'Revenue']].agg({
    'count',
    'mean',
    'median',
    'std',
    'var',
    'min',
    'max',
    'skew',
    'kurt'
}).T # summary = summary. transpose()

summary 
```

##### **7.1 Clean the Quantity and Rating Variables**

```python
df.loc[df['Rating'] < 0, 'Rating'] = np.nan
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
```

```python
df.loc[df['Quantity'] < 0, 'Quantity'] = np.nan
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
summary = df[['Quantity', 'UnitPrice', 'Discount', 'Rating', 'Revenue']].agg({
    'count',
    'mean',
    'median',
    'std',
    'var',
    'min',
    'max',
    'skew',
    'kurt'
}).T 

summary 
```
